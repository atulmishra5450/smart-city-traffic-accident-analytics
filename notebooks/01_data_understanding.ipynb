# 01 — Data Understanding
## Smart City Traffic & Accident Analytics

Inspect the raw dataset **without** modifying it.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/raw/smart_city_traffic_accident_messy_350k.csv')
print(f'Shape: {df.shape}')
```

```python
print('=== Data Types ===')
print(df.dtypes)
```

```python
print('=== Missing Values ===')
m = df.isnull().sum()
print(m[m > 0])
```

```python
print('=== Descriptive Statistics ===')
df.describe()
```

```python
# Unique values per categorical column
for col in ['City','Weather','Road_Condition',
            'Accident_Severity','Traffic_Signal_Status']:
    print(f'{col}: {sorted(df[col].dropna().unique().tolist())}')
```

```python
# City distribution
fig, ax = plt.subplots(figsize=(9,4))
df['City'].value_counts().plot(kind='bar', ax=ax, color='#3b82d4')
ax.set_title('Records per City')
plt.tight_layout()
plt.show()
```

```python
# Speed range check
print('Speed range:', df['Avg_Speed_kmph'].min(), '-', df['Avg_Speed_kmph'].max())
print('Invalid (<1 or >150):', ((df['Avg_Speed_kmph'] <= 0) | (df['Avg_Speed_kmph'] > 150)).sum())
```

```python
# Congestion check
print('Congestion range:', df['Congestion_Level_%'].min(), '-', df['Congestion_Level_%'].max())
print('Invalid (<0 or >100):', ((df['Congestion_Level_%'] < 0) | (df['Congestion_Level_%'] > 100)).sum())
```

```python
# Temperature outlier check (IQR)
q1, q3 = df['Temperature_C'].quantile([0.25, 0.75])
iqr = q3 - q1
print(f'IQR bounds: [{q1 - 1.5*iqr:.1f}, {q3 + 1.5*iqr:.1f}]')
print('Outliers:', ((df['Temperature_C'] < q1-1.5*iqr) | (df['Temperature_C'] > q3+1.5*iqr)).sum())
```

```python
# Accident distribution
print(df['Accident_Severity'].value_counts(dropna=False))
```
