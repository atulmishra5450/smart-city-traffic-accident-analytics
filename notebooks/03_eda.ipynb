# 03 — Exploratory Data Analysis
## Smart City Traffic & Accident Analytics

20 visualisations covering traffic, congestion, accidents, weather, and emergency response.

```python
import sys
sys.path.insert(0, '../python')
from utils import set_style, PROCESSED_DATA
import pandas as pd
set_style()
df = pd.read_csv(PROCESSED_DATA / 'traffic_processed.csv')
print(df.shape)
```

### 1. Traffic Volume by City

```python
import matplotlib.pyplot as plt
city_vol = df.groupby('City')['Vehicle_Count'].sum().sort_values(ascending=True) / 1e6
fig, ax = plt.subplots(figsize=(9,5))
city_vol.plot(kind='barh', ax=ax, color='#3b82d4')
ax.set_xlabel('Total Vehicles (M)')
ax.set_title('Total Traffic Volume by City')
plt.tight_layout()
plt.show()
```

### 2. Hourly Traffic Pattern

```python
hourly = df.groupby('Hour')['Vehicle_Count'].mean()
fig, ax = plt.subplots(figsize=(10,4))
ax.plot(hourly.index, hourly.values, color='#3b82d4', linewidth=2, marker='o', ms=4)
ax.fill_between(hourly.index, hourly.values, alpha=0.15, color='#3b82d4')
ax.set_title('Average Traffic Volume by Hour')
ax.set_xticks(range(0,24))
plt.show()
```

### 3. Accident Severity Distribution

```python
acc = df[df['Accident_Flag']==1]
sev = acc['Accident_Severity'].value_counts().reindex(['Minor','Moderate','Severe','Fatal'])
colors = ['#86efac','#fbbf24','#f97316','#ef4444']
fig, ax = plt.subplots(figsize=(7,4))
sev.plot(kind='bar', ax=ax, color=colors, rot=0)
ax.set_title('Accident Severity Distribution')
plt.show()
```

### 4. Weather vs Accident Rate

```python
w_acc = df.groupby('Weather').agg(total=('Accident_Flag','count'), acc=('Accident_Flag','sum')).assign(rate=lambda x: x['acc']/x['total']*100).sort_values('rate', ascending=False)
fig, ax = plt.subplots(figsize=(9,4))
ax.bar(w_acc.index, w_acc['rate'], color='#f97316')
ax.set_title('Accident Rate by Weather')
ax.set_xticklabels(w_acc.index, rotation=30)
plt.show()
```

### 5. Emergency Response by City

```python
acc = df[df['Accident_Flag']==1]
ert = acc.groupby('City')['Emergency_Response_Min'].mean().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(9,5))
ert.plot(kind='bar', ax=ax, color='#7c5cd8', rot=30)
ax.axhline(15, color='red', linestyle='--', label='15-min target')
ax.set_title('Avg Emergency Response by City')
ax.legend()
plt.show()
```
