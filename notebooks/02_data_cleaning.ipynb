# 02 — Data Cleaning
## Smart City Traffic & Accident Analytics

Complete cleaning pipeline. All steps are modular and documented.

```python
import sys
sys.path.insert(0, '../python')
import pandas as pd
from utils import load_raw, save_cleaned
from data_cleaning import clean_pipeline

raw = load_raw()
print(f'Raw shape: {raw.shape}')
```

### Step 1: Remove Duplicates

```python
from data_cleaning import remove_duplicates
df = remove_duplicates(raw.copy())
print(f'After dedup: {df.shape}')
```

### Step 2: Handle Missing Values

```python
from data_cleaning import handle_missing
df = handle_missing(df)
print('Missing after imputation:')
print(df.isnull().sum()[df.isnull().sum()>0])
```

### Step 3: Normalise Categories

```python
from data_cleaning import normalise_categories
df = normalise_categories(df)
print('Weather unique:', df['Weather'].unique())
print('Road unique:', df['Road_Condition'].unique())
```

### Step 4: Fix Invalid Speed

```python
from data_cleaning import fix_speed
df = fix_speed(df)
print('Speed range:', df['Avg_Speed_kmph'].min(), '-', df['Avg_Speed_kmph'].max())
```

### Step 5: Fix Invalid Congestion

```python
from data_cleaning import fix_congestion
df = fix_congestion(df)
print('Congestion range:', df['Congestion_Level_%'].min(), '-', df['Congestion_Level_%'].max())
```

### Step 6: Fix Temperature Outliers

```python
from data_cleaning import fix_temperature
df = fix_temperature(df)
print('Temp range:', df['Temperature_C'].min(), '-', df['Temperature_C'].max())
```

### Step 7: Parse Datetime

```python
from data_cleaning import parse_datetime
df = parse_datetime(df)
print(df[['Datetime','Year','Month','Hour','Peak_Hour_Flag']].head())
```

### Step 8 & 9: Accident Flag + Severity Score

```python
from data_cleaning import create_accident_flag, create_severity_score
df = create_accident_flag(df)
df = create_severity_score(df)
print('Accidents:', df['Accident_Flag'].sum())
print('Severity score distribution:')
print(df['Severity_Score'].value_counts().sort_index())
```

```python
save_cleaned(df)
print('Cleaned dataset saved.')
```
